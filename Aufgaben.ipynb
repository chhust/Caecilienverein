## FUNDSTELLEN ANSEHEN

`.splitlines()` zerlegt den Text in eine Liste von Zeilen.

+ Mit `for` "iterieren" wir diese Zeilen, d. h., wir gehen die Einzelelemente dieser Liste einzeln durch.
+ Mit `if` können wir eine Entscheidung treffen und etwas nur dann machen lassen, wenn eine Bedingung zutrifft.

Die Einrückung gehört zur Syntax der Sprache: Sie zeigt, was innerhalb der Schleife und der Bedingung passiert.

Das sehen wir uns erst mal an einem abstrakten Beispiel an:

```python
demo = ["Element 1", "zweites Element", "DREI", "ElEmEnT vIeR"]

print(demo)             # die ganze Liste zeigen
print()

for eintrag in demo:   # die Listenelemente einzeln zeigen
    print(eintrag)
print()

for eintrag in demo:    # nochmals alle iterieren, ...
    if "Element" in eintrag:    # aber nur die anzeigen, die das Wort "Element" enthalten
        print(eintrag)
    else:
        print("Nope!")
```

## APPENDIX A:
## Suchen mit Mustern (reguläre Ausdrücke)

Bisher haben wir nach festen Zeichenfolgen gesucht. "Reguläre Ausdrücke" ("Regex" für *regular expressions*) suchen stattdessen nach einem Muster.

Python hat dafür das Modul `re`.

`re.findall(muster, text)` gibt eine Liste aller Treffer zurück; `len(...)` zählt sie.

| Zeichen | bedeutet | Beispiel |
|---|---|---|
| `.` | genau ein beliebiges Zeichen | `Kornm.ller` |
| `[ ]` | eines der Zeichen in der Klammer | `[Ff]ür die Aufnahme` |
| `\|` | oder | `Haberl\|Witt` |
| `\d` | eine Ziffer | `\d+ ℳ` |
| `+` · `{1,3}` | einmal oder öfter · ein- bis dreimal | `Ko.{1,3}nm` |

Muster werden als sogenannte *raw strings* geschrieben: `r"..."` , damit Python den Backslash `\` nicht selbst verarbeitet.

```python
import re

treffer = re.findall(r"Kornm.ller", band_alt)
len(treffer), set(treffer)
```
